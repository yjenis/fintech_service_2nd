```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
from datetime import datetime, timedelta
import pandas as pd
import time
from tqdm import tqdm

options = Options()
options.add_experimental_option("detach", True)
options.add_argument("start-maximized")
options.add_argument("Chrome/135.0.0.0")
options.add_argument("lang=ko_KR")
# 웹브라우저가 백그라운드에서 작동하도록 설정
options.add_argument("--headless")
options.add_argument("--no-sandbox")
options.add_argument("--disable-dev-shm-usage")

driver = webdriver.Chrome(
    service=Service(ChromeDriverManager().install()),
    options=options
    )

url = "https://play.google.com/store/apps/details?id=viva.republica.toss"
driver.get(url)

# 평점 및 리뷰 옆의 -> 버튼 클릭
wait = WebDriverWait(driver, 10)
button = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, "#yDmH0d > c-wiz.SSPGKf.Czez9d > div > div > div:nth-child(1) > div > div.wkMJlb.YWi3ub > div > div.qZmL0 > div:nth-child(1) > c-wiz:nth-child(5) > section > header > div > div:nth-child(2) > button")))
button.click()

# 최신순으로 리뷰를 정렬하기 위해서 버튼 클릭
wait = WebDriverWait(driver, 10)
button = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, "#sortBy_1")))
button.click()

# 최신을 찾아 클릭
wait = WebDriverWait(driver, 10)
button = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, "span[aria-label*='최신']")))
button.click()

time.sleep(5)

# 오늘 날짜를 추출하고 2년 전 날짜를 계산해 반복문 조건으로 활용
today = datetime.today()
two_yr_ago = today - timedelta(days=60)

all_reviews = ""
while True:
    try:
        # 리뷰가 담긴 창을 찾아서 JavaScript로 1000px씩 아래로 스크롤
        driver.execute_script("document.querySelector('.fysCi.Vk3ZVd').scrollBy(0, 1000)")
        time.sleep(1)

        # 리뷰일
        review_date = driver.find_elements(By.CSS_SELECTOR, "div.Jx4nYe > span.bp9Aid")[-1].text.replace("년 ", "-").replace("월 ", "-").replace("일", "")
        review_date = datetime.strptime(review_date, "%Y-%m-%d")
        all_reviews = driver.find_elements(By.CSS_SELECTOR, ".RHo1pe")
        print(review_date, len(driver.find_elements(By.CSS_SELECTOR, ".RHo1pe")), end="\r")
    except:
        print(review_date, len(driver.find_elements(By.CSS_SELECTOR, ".RHo1pe")), end="\r")
        all_reviews = driver.find_elements(By.CSS_SELECTOR, ".RHo1pe")
        
#     if review_date < two_yr_ago:
#         break

########################################################################
# 리뷰 데이터 추출

all_reviews = driver.find_elements(By.CSS_SELECTOR, ".RHo1pe")
result = []
for review in tqdm(all_reviews):
    review_date = review.find_elements(By.CSS_SELECTOR, "div.Jx4nYe > span.bp9Aid")[-1].text.replace("년 ", "-").replace("월 ", "-").replace("일", "")
    review_date = datetime.strptime(review_date, "%Y-%m-%d")
    # 별점
    rating = review.find_element(By.CSS_SELECTOR, "div.Jx4nYe > div").get_attribute("aria-label").split()[3][0]
    # 사용자리뷰
    user_review = review.find_element(By.CSS_SELECTOR, "div.h3YV2d").text
    # 회사 답변
    try:
        reply = review.find_element(By.CSS_SELECTOR, "div.ocpBU div.ras4vb > div").text #.get_attribute("innerHTML")
    except:
        reply = None
    print(review_date, rating, user_review, reply)
    result.append((review_date, rating, user_review, reply))

result_df = pd.DataFrame(result, columns=["리뷰일", "평점", "사용자리뷰", "개발자답변"])    
result_df
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from webdriver_manager.chrome import ChromeDriverManager
from datetime import datetime, timedelta
import pandas as pd
import time
from tqdm import tqdm

SCROLL_PAUSE_TIME = 2
scroll_height = 1400
scroll_times = 30  # 필요에 따라 조절

for i in range(scroll_times):
    driver.execute_script(f"window.scrollTo(0, {scroll_height * (i + 1)})")
    time.sleep(SCROLL_PAUSE_TIME)

all_reviews = driver.find_elements(By.CSS_SELECTOR, ".RHo1pe")
result = []

for review in tqdm(all_reviews):
    review_date = review.find_elements(By.CSS_SELECTOR, "div.Jx4nYe > span.bp9Aid")[-1].text.replace("년 ", "-").replace("월 ", "-").replace("일", "")
    review_date = datetime.strptime(review_date, "%Y-%m-%d")
    # 별점
    rating = review.find_element(By.CSS_SELECTOR, "div.Jx4nYe > div").get_attribute("aria-label").split()[3][0]
    # 사용자리뷰
    user_review = review.find_element(By.CSS_SELECTOR, "div.h3YV2d").text
    # 회사 답변
    try:
        reply = review.find_element(By.CSS_SELECTOR, "div.ocpBU div.ras4vb > div").text #.get_attribute("innerHTML")
    except:
        reply = None
    print(review_date, rating, user_review, reply)
    result.append((review_date, rating, user_review, reply))

result_df = pd.DataFrame(result, columns=["리뷰일", "평점", "사용자리뷰", "개발자답변"])    
result_df
```
