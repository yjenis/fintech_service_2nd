```python
import os
from dotenv import load_dotenv
import openai

# .env 파일에서 API 키 로드
load_dotenv()
apikey = os.getenv('OPENAI_API_KEY')  # 환경 변수명은 정확히 확인하세요

# API 키 설정
openai.api_key = apikey

# ChatCompletion 요청
response = openai.ChatCompletion.create(
    model="gpt-4o",
    messages=[
        {"role": "user", "content": "Write a one-sentence bedtime story about a unicorn."}
    ]
)

# 결과 출력
print(response['choices'][0]['message']['content'])
```

```python
import os
from dotenv import load_dotenv
from openai import OpenAI

# .env에서 API 키 불러오기
load_dotenv()
apikey = os.getenv('OPENAI_API_KEY')  # 정확한 환경 변수명 확인

# 클라이언트 객체 생성
client = OpenAI(api_key=apikey)

# ChatCompletion 요청
response = client.chat.completions.create(
    model="gpt-4o",
    messages=[
        {"role": "user", "content": "Write a one-sentence bedtime story about a unicorn."}
    ]
)

# 응답 출력
print(response.choices[0].message.content)
```
