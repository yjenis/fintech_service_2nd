```python
import gradio as gr
from google import genai

# 🔑 여기에 본인의 API 키를 입력하거나 안전하게 관리
client = genai.Client(api_key=apikey)

# 🌐 대화 함수
def chat_with_gemini(prompt):
    try:
        response = client.models.generate_content(
            model="gemini-2.0-flash",
            contents=prompt,
        )
        return response.text
    except Exception as e:
        return f"[에러 발생] {str(e)}"

# 🧩 Gradio 인터페이스 정의
iface = gr.Interface(
    fn=chat_with_gemini,
    inputs=gr.Textbox(label="질문을 입력하세요", lines=4, placeholder="예: 오늘 날씨 어때?"),
    outputs=gr.Textbox(label="Gemini 응답", lines=6),
    title="Gemini 2.0 Flash Chat",
    description="Google AI Gemini-2.0-Flash 모델을 이용한 간단한 채팅 서비스입니다."
)

iface.launch()
```
