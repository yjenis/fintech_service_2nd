국민은행 앱 사용자 리뷰 데이터 JSON 파일을 읽어와서 csv파일로 변환하기

```python
import json
import pandas as pd
```

```python
with open("./data/com_kbstar_kbbank.json", 'r', encoding="utf-8") as file:
    data = file.read()
```

```python
data[:100]
```

```python
reviews = json.loads(data)
len(reviews)
```

```python
reviews[0]
```

```python
result = {}
for key in reviews[0]:
    temp =  []
    for review in reviews:
        temp.append(review[key])
    result[key] = temp
len(result['별점'])
```

```python
df = pd.DataFrame(result)
df
```

별점은 숫자로 변환해서 넣기, 나머지는 특수문자 제거하기

```python
for key in reviews[0]:
    for review in reviews:
        print(review[key].replace("?", ""))
```

```python
result = {}
for key in reviews[0]:
    temp =  []
    for review in reviews:
        if key == '별점':
            temp.append(int(review[key]))
        else:
#             print(type(review[key]))
            temp2 = str(review[key]).replace("?", " ").replace("(", " ").replace(")", " ").replace("!", " ")
            temp.append(temp2)
    result[key] = temp
len(result['별점'])
df = pd.DataFrame(result)
df
```

```python
df.info()
```

```python
result2 = {}
for review in reviews:
    for key, value in review.items():
        result2.setdefault(key, []).append(value)
result2_df = pd.DataFrame(result2)
# result2_df
```

```python
with open("./data/전국치킨집리스트.json", 'r', encoding="utf-8") as file2:
    chick = file2.read()
data = json.loads(chick)
```

```python
len(data)
```

* data에는 리스트 요소가 37개 있슴.
* 리스트 요소 1개에는 1000개의 딕셔너리가 있고 그 안에 가게 정보가 있슴
* 가게 정보는 data[0]['body']['items'] 에 있다.
* 가게 정보를 구성하는 key는 39개이다.

```python
data[0]['body']['items'][0]
```

```python
result3 = {}
for page in data:
    for item in page['body']['items']:
        for key, value in item.items():
            result3.setdefault(key, []).append(value)
#         print(key, value)
len(result3['bizesId'])
```

```python
df = pd.DataFrame(result3)
df
```

```python
data[0]['header']['columns']
```

```python
result4 = {}
for page in data:
    for item in page['body']['items']:
        for key, value in zip(page['header']['columns'], item.values()):
            result4.setdefault(key, []).append(value)

df2 = pd.DataFrame(result4)
df2
```
